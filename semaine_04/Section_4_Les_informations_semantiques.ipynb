## Description du problème

#### Description du problème

**Description du problème**

On se place dans un contexte qui est celui d'un patient qui arrive chez son médecin. Il porte une boîte noire qui fournit au médecin des mesures brutes de type (t 37).



Le médecin est un humain. Il est donc doué d'une capacité d'analyse et de raisonnement qui lui permet, d'après

- la valeur 37,
- l'entête t,
- la position de la boîte noire,
- sa localisation européenne,

d'en déduire que cette donnée brute est la mesure de température du patient exprimée en degré Celsius.

> > > 

Une machine (*i.e. un équipement, un processus, une application...*) ne peut pas traiter cette donnée brute car il lui manque :

> 1. la signification de ce qu'est cette donnée **=>** C'est ce qu'on appelle une métadonnée. Dans notre exemple, il lui manque le fait que 37 est une valeur réelle de température dont l'acronyme est t et l'unité degré Celsius ;
> 2. le contexte de cette donnée brute **=>** Cette extension de la métadonnée est appelée une ontologie. Dans notre exemple, ce contexte pourrait être le capteur qui fournit  cette donnée, les conditions de mesure, la localisation du capteur, le patient qui le porte, la pièce et l'environnement où est faite la mesure...

L'utilisation de métadonnées unifiées est indispensable pour gérer l'hétérogénéité et l'interopérabilité des données échangées entre objets connectés et applications IoT.  En effet, ces métadonnées permettent à n'importe quel objet, équipement, processus, application et utilisateur final de comprendre et d'analyser une donnée de la même manière. Les métadonnées permettent aussi d'introduire un premier niveau de connaissance permettant une analyse de donnée sémantique (*semantic data analytics*) simple. Dans l'exemple médecin-patient précédemment introduit, l'introduction de seuils sur les métadonnées permet par exemple d'envoyer une alarme au médecin si la température du patient dépasse un certain seuil (*fièvre*).

L'ontologie nous fournit une sémantique enrichie, au sein d'un domaine d'application donné,  sur l’environnement autour de nos données et de nos objets connectés. Sa mise en œuvre est indispensable car elle permet de gérer l'hétérogénéité globale des données et des informations d'un environnement et d'un domaine d'application donnés (*elle rend les données interopérables*). Une ontologie modélise donc un environnement en terme de concepts  le décrivant et de liens entre ces concepts. Dans notre exemple médecin-patient, la connaissance sémantique enrichie apportée par l'usage d'une ontologie (*i.e. contexte de la donnée brute introduit dans l'exemple médecin-patient*) permet déjà l'implémentation de mécanismes de traitement sémantique évolués de nos données (*semantic data analytics*), par exemple : détection de fautes, de défauts de fonctionnement ou de données erronées, détection de similarités ou encore détection de patients colocalisés dans le même immeuble et qui ont les mêmes valeurs critiques de mesures vitales (*donc la même pathologie*)...

La spécification, la formalisation et la mise en œuvre d'une ontologie permet de résoudre l'interopérabilité au niveau données/informationnel pour l'internet des objets. De plus et si cette ontologie est standardisée et partagée par les acteurs de plusieurs domaines d'application (*concepts unifiés et partagés entre différents verticaux*), elle permettra aussi d'aborder des scénarios multi-domaines. Mais ceci n'est pas suffisant car, comme indiqué dans la présentation de la semaine 4, il est aussi important de résoudre la problématique d'interopérabilité d'une manière plus globale, y compris au niveau équipements (*devices* ou objets connectés), réseau et système. Pour cela, une solution couramment utilisée est de gérer cette interopérabilité au niveau sémantique : on parle alors d'interopérabilité sémantique.

L'interopérabilité sémantique consiste à modéliser sémantiquement, à l'aide d'une ontologie étendue, toutes les interactions entre les entités (*devices* et/ou objets connectés) et les systèmes qui doivent interagir entre eux. Cette interopérabilité sémantique concerne tous les niveaux, y compris ceux précédemment mentionnés, c'est-à-dire : données/informationnel, entité/device/objet connecté, réseau, système et application. Pour la mettre en œuvre, il est courant d'étendre une ontologie de base (modélisant un environnement donné) avec des concepts de plus haut niveau, par exemple au niveau service/application via un modèle de service. En effet, si on modélise

* un service d'une manière unifiée au travers d'une ontologie générique de service dédiée, notamment en terme de : '*Service profile*' (ce qu'est, ce que fait le service), '*Service Process*'  (comment le service fonctionne, par exemple comment l'invoquer ou modifier un de ses paramètres de fonctionnement) et 'Service Grounding' (comment on accède au service, par exemple via quel protocole réseau),
* un capteur comme un service de mesure fournissant une mesure,

alors, quel que soit ce capteur et quel que soit son domaine d'application (eSanté, maison ou ville intelligente, système de transport intelligent*...*), il est toujours vu et décrit comme un service générique de mesure à l'aide d'une ontologie de service unifiée et commune/partagée, et donc on sait comment le gérer et comment récupérer la donnée de mesure qu'il fournit.

## Spécification d'une ontologie

#### Spécification d'une ontologie

Spécifier une ontologie et sa sémantique, c’est décrire les concepts (*c'*est-à-dire les objets) qu'elle doit modéliser pour un environnement et un domaine d'application donné, ainsi que les propriétés de ces objets/concepts. Les propriétés d'un objet/concept sont de deux types :

1. Les données/attributs qui caractérisent/décrivent l'objet (*i.e.* le concept). On parle alors de propriétés de donnée ou '*data properties*',
2. Les relations entre les objets/concepts. On parle alors de propriétés d'objet ou '*object properties'*.



On prend l’exemple simplifié d’une voiture. C’est un véhicule qui est composé : d'une caisse, d'un châssis et d'un moteur. On vient ainsi de définir notre premier concept, la voiture, avec trois autres concepts, caisse/châssis/moteur qui sont liés à la voiture par la relation ‘est composé de', une première '*object properties*' de la classe voiture. Il ne nous reste plus qu’à finir de décrire et de modéliser ces concepts avec toutes leurs propriétés.

Pour modéliser nos concepts (*i.e.* nos objets réels) et leurs propriétés, on utilise un modèle de description. Le modèle le plus utilisé pour modéliser et spécifier une ontologie est le modèle de classes d'UML (*Unified Modeling Language*), même si parfois un graphe peut-être utilisé. Dans le modèle de classes d'UML appliqué à la modélisation/spécification d'une ontologie, on a les conventions suivantes :

> * nos concepts (*i.e.* nos objets) sont représentés par des classes, indiquées par des rectangles dans le diagramme UML. Ces classes sont les nœuds du graphe UML et ont pour label l'identifiant des concepts qu'elles modélisent (par exemple le nom du concept) ;
> * les flèches pleines terminées par un triangle représentent une relation de type sous-classe entre deux classes.  L'origine de la flèche est la sous-classe de la classe se situant à la destination de la flèche. Si on reprend notre exemple de voiture, on à la relation suivante : une voiture est une sous-classe de la classe générique véhicule (voir le diagramme UML d'une voiture *- 'Car UML Model' -* présenté ci-après) ;
> * les '*object properties*' entre deux classes (c'est-à-dire les relations entre deux objets/concepts) sont matérialisées par des flèches en pointillés, des arêtes du diagramme UML, étiquetées avec le type de la relation. Une '*object properties*' est toujours instanciée entre la classe à l'origine de la flèche et la classe à la destination de la flèche. Dans notre exemple, une voiture est notamment composée d'un moteur (*'isComposedOf',* une première *'object properties'* de la classe voiture). Cette relation est matérialisée dans le diagramme UML d'une voiture par une flèche pointillée : partant de la classe voiture et se terminant à la classe moteur, labellisée '*isComposedOf*' (voir le diagramme UML d'une voiture *- 'Car UML Model' -* présenté ci-après) ;
> * les '*data properties*', c'est-à-dire les attributs qui décrivent l'objet/ le concept,  sont matérialisées par des rectangles en pointillé.

Le nœud du graphe UML modélisant un concept d'une ontologie a donc la forme suivante :

UML Class Representation

Si nous revenons à notre voiture, c’est un véhicule, donc la racine de notre modèle UML est logiquement  la classe véhicule.

Car UML Model

Si nous regardons l'arête qui relit la classe  véhicule, nous pouvons dire qu’un véhicule a un sous-type voiture (*une voiture est une sous-classe de véhicule*), c’est ce que nous avons rajouté dans notre diagramme UML. Cette voiture est aussi, en général, la propriété d’un conducteur, d’où le lien « à un propriétaire » (*i.e.* le lien *'asOwner'*) avec une classe conducteur, décrite notamment par le nom du conducteur.

Notre voiture est en particulier décrite par :

> * son constructeur,
> * son n° de série,
> * son type,
> * son modèle,

et ces attributs sont rajoutés sous le nom du nœud ‘*Car*’ dans notre graphe. Notre voiture est elle-même composée d’une caisse, d’un châssis et d’un moteur (matérialisé par les liens *'isComposedOf'* sur le diagram UML d'un voiture présenté ci-dessus).

Vous retrouvez donc logiquement ces concepts dans votre graphe simplifié comme nœuds du diagramme UML de l'ontologie voiture présenté ci-dessus.

#### Mongo DB

### MongoDB

MongoDB est une base de données qui permet de stocker des données formatées en JSON. Contrairement aux bases de données SQL, les données ne suivent pas un formatage strict avec un certain nombre de champs prédéfinis.  MongoDB définit deux niveaux de hiérarchie : la base de données (*database*) et la collection. Une collection peut être vue comme un ensemble de structure JSON et la base de données comme un regroupement de collections. Il est possible d'insérer, de rechercher ou d'effacer des éléments dans une collection.

Le module pymongo permet de gérer le dialogue entre un programme Python et une base MongoDB. L'import doit être présent au début du programme.

*MyOntoSens* est une ontologie standardisée par le groupe de l'ETSI travaillant sur les réseaux personnels (*Body Area Network*), mais peut s'étendre à n'importe quel réseau de capteurs ainsi qu'aux réseaux LPWAN. Elle est définie dans la spécification technique [TS 103 378](https://www.etsi.org/deliver/etsi_ts/103300_103399/103378/01.01.01_60/ts_103378v010101p.pdf).

![node class diagram](/static/class_node_2_.png)

La figure ci-dessus représente les concepts du modèle sémantique d'un nœud dans le réseau de capteur ; c'est-à dire un Pycom dans notre cas. Nous allons nous intéresser plus particulièrement au processus de remontée de l'information. Le nœud possède un capteur (*sensor*) qui va remonter des mesures (*measurements*). Vous pouvez facilement étendre ce modèle sémantique à des actionneurs (*actuators*). Le document prévoit aussi le cas des réseaux maillés (*Mesh*) où un nœud peut être un puits (*sink*) par lequel transite des informations.

#### Classe *measurement*

Dans la première partie précédente de ce TP, nous avons stocké dans une base de données MongoDB les données brutes remontant du Pycom. Il est très dur de les exploiter si on n'a pas une connaissance précise du format des données envoyées. Par exemple :

```
{ "_id" : ObjectId("5bfa563aec3305156374359d"), "measure" : [ 2004, 100816, 0 ] }

Ainsi, au lieu de mettre un tableau, on pourrait mettre un dictionnaire avec les différentes mesures possibles :

Mais ce n'est pas suffisant car si d'autres personnes utilisent les mêmes identifiants et si la précision n'est pas la même (ex : la valeur mesurée a été multipliée par 100 dans un groupe et 1000 dans l'autre), les données ne seront pas interopérables.

Il faut rendre unique ces identifiants et être certain que personne d'autre au monde ne pourra les utiliser. Avez-vous une idée de ce qui pourrait nous permettre d'obtenir cette unicité ?

#### JSON-LD

#### JSON-LD

[JSON-LD](https://www.w3.org/TR/2014/REC-json-ld-20140116/) (pour *Linked Data*) est défini par le W3C. Il introduit des mots-clés commençant par le caractère @ qui vont aider à structurer les informations et utiliser des modèles de données communs. C'est le langage de description qui va être utilisé pour formaliser notre ontologie.

Nous avons vu précédemment que le mot-clé temperature n'est pas suffisant pour indiquer comment sera représentée la température. En effet, lors des séances précédentes, vous avez pu modifier les programmes et adopter une représentation qui vous est propre. Comme temperature est un mot commun, plusieurs développeurs ont dû l'employer. Il faut rendre cet identifiant unique de manière à lever l'ambiguïté ; par exemple en concaténant une URI unique *http://user.ackl.io/devEUI* où *devEUI* est l'identifiant unique de votre Pycom.

On aura donc :

```
{"http://user.ackl.io/70b3d54994c61237/temperature" : 2004, "http://user.ackl.io/70b3d54994c61237/pression": 100816, "http://user.ackl.io/70b3d54994c61237/humidity" : 0}
```

qui représentera les données mesurées par un Pycom et :

```
{"http://user.ackl.io/70b3d5499aef7aca/temperature" : 20043, "http://user.ackl.io/70b3d5499aef7aca/pression": 1008146, "http://user.ackl.io/70b3d5499aef7aca/humidity" : 7810}
```

qui servira pour identifier des mesures faites par un autre équipement.

Il n'est pas nécessaire que cette URI soit une URL valable (c'est-à-dire qu'en la recopiant dans un navigateur, on obtienne un résultat). Mais c'est une bonne pratique de le faire car cela permet de connaître le format des données ; par exemple si votre URI pointe vers une page qui donne la structure JSON-LD des données.

#### JSON-LD suite

Bien entendu, cette représentation n'est pas optimale car l'URI est relativement longue. Le mot clé @context de JSON-LD permet de représenter une seule fois cette URI.

Le mot clé @context permet de définir le vocabulaire utilisé dans l'enregistrement. Ainsi :

```
"@context": "http://user.ackl.io/70b3d54994c61237",
"temperature" : 2004,
"pression": 100816,
"humidity" : 0
}
```

permet de rendre unique les définitions de temperature, pression et humidity en les associant une seule fois à l'URI complète. Le contexte permet aussi de préciser les différents termes :

```
{
"@context": {
"temperature": {
"@id": "http://user.ackl.io/70b3d54994c61237/temperature"
},
"pression" : {
"@id": "http://user.ackl.io/70b3d54994c61237/pression"
},
"humidite" : {
"@id" :"http://user.ackl.io/70b3d54994c61237/temperature"
}
},
"temperature" : 2004,
"pression": 100816,
"humidity" : 0
}
```

Ici, on définit les termes temperature, pression et humidity, et une propriété qui est l'identifiant unique (@id) est associée à chacun de ces mots. @id indique que la valeur est définie sous forme d'URI.

Cela semble peu efficace par rapport aux précédentes représentations mais la partie contexte peut être regroupée dans son propre objet.

Le programme ci-dessus (mongo_aggregate.py) illustre l'utilisation de l'agrégation sur un cas simple. Après avoir cherché l'identifiant de l'objet dont le nom est stocké dans la variable sensor_location (ici Room A123 qui est la valeur par défaut), la fonction aggregate est appliquée à la collection des mesures.

```
res = measures.aggregate([{"$match" : {"SensorCharacteristics" : sensor_id }}])

La syntaxe est un peu obscure car il y a beaucoup d'imbrications de structures, de parenthèses, d'accolades, de crochets et de guillemets. On va décortiquer tout ça :

* Aggregate prend un argument qui est un tableau qui va contenir un ensemble d'étapes d'agrégation ;
* une étape d'agrégation est elle-même représentée par une liste (dictionnaire) contenant un mot-clé, représenté par un ou plusieurs opérateurs (repérables car précédés du signe $) et des instructions associées à cet opérateur.

Dans l'exemple précédent, l'étape d'agrégation consiste à choisir, dans la base de données, les enregistrements où l'entrée SensorCharacteristics correspond à l'identifiant du capteur.

#### Opérations pendant l'agrégation

L'intérêt de l’agrégation n'est pas uniquement de sélectionner des enregistrements mais également de fournir un résultat qui va regrouper les données ; d'où le nom d’agrégation. La syntaxe est dans la continuation de ce que nous avons vu précédemment. Dans le tableau fourni en argument à la fonction aggregate , d'autres opérateurs peuvent être utilisés.

Modifiez  le programme mongo_aggregate.py, en ajoutant la commande d'agrégation suivante et en affichant le résultat:

```
res = measure.aggregate([
{"$match" : {"SensorCharacteristics" : sensor_id }},
{"$group" : {
"_id": None, # equivalent to null in MongoDB
"count": {"$sum": 1},
"temp_min": {"$min": "$Temperature"},
"temp_max": {"$max": "$Temperature"}
}
}
])
```

On continue à filtrer les enregistrements en fonction du champ *SensorCharacteristics* et on applique un traitement en regroupant les informations (opérateur $group). Cet opérateur va créer un nouvel enregistrement comportant 4 champs :

* _id : indispendable pour MongoDB. Il indique l'enregistrement dans lequel nous voulons regrouper l'information.  Comme nous n'avons qu'un regroupement, il peut valoir None (ce qui est équivant à null dans la syntaxe de MongoDB) ;
* count : va servir à indiquer le nombre d'enregistrements de ce type. À chaque fois qu'un enregistrement est trouvé, la valeur 1 est ajoutée au champ (opérateur $sum) ;
* temp_min et temp_max fonctionnent suivant le même principe avec les opérateurs $min et $max. Par contre, ces opérateurs fonctionnent avec les champs de l'enregistrement trouvé. $Temperature fait référence à Temperature de l'enregistrement initial.

{'count': 1000, '_id': None, 'temp_min': -4.99, 'temp_max': 30.0}

#### Regroupement

Dans l'exemple précédent, l'étape $match a permis d'isoler les données émises par un capteur. On peut également interroger l'intégralité de la base et regrouper les résultats par capteur en utilisant le champ _id qui contiendra la valeur du champ SensorCharacteristics.

```
res = measure.aggregate([
{"$match": {"@context":"http://user.ackl.io/schema/BME280"}},
{"$group" : {
"_id": "$SensorCharacteristics",
"count": {"$sum": 1},
"temp_min": {"$min": "$Temperature"},
"temp_max": {"$max": "$Temperature"},
"avg_temp": {"$avg": "$Temperature"},
"avg_pres": {"$avg": "$Pressure"},
"avg_humi": {"$avg": "$Humidity"}
}
}
])
```

Il est également possible de mettre des opérateurs dans une valeur. Par exemple :

```
res = measure.aggregate([
{"$match": {"$and": [
{"@context":"http://user.ackl.io/schema/BME280"},
{"Temperature": {"$lte": 10, "$gt": 0 }}]}},
{"$group": {
"_id" : None,
"count": {"$sum": 1}}}
])
```

La correspondance (match) se fait sur les champs @context et Temperature grâce à l'opérateur $and qui prend comme valeur un tableau indiquant les champs à tester. Pour Temperature, le test porte sur les valeurs plus petites ou égales ($lte: *less than or equal*)  à 10 et strictement plus grandes que 0 ($gt: *greater than*).

L'ensemble des opérateurs est disponible [sur ce site](https://docs.mongodb.com/manual/reference/operator/query/), et va vous aider à répondre aux questions suivantes.

### 5)	MongoDB : le langage de requête sur des données JSON

### **MongoDB : le langage de requête sur des données JSON**

La fonction de recherche sur une collection de données JSON, comme celle de vos objets connectés de type ‘Things’, est *find()*. Elle s’utilise comme suit :

*db.collection.find(query, projection)*

où ‘query’ permet de spécifier les filtres de sélection/recherche et ‘projection’ permet de spécifier les champs des éléments de votre collection qui doivent être retournés. Si 'projection' est omis, tous les champs sont retournés.

La fonction find retourne une liste Python de tous les éléments d'une collection (*ici 'db'*) qui correspondent aux filtres de selection/recherche (*i.e. le paramettre query*)

Une autre fonction de recherche, find_one, a aussi été introduite afin de ne retourner que le premier élément trouvé  d'une collection (*ici 'db'*) qui correspond aux filtres de selection/recherche (*i.e. le paramettre query*). Elle s'utilise de la même manière que la fonction find.

Prenons maintenant l’exemple d’une base de données MongoDB de test (de nom testDB) qui contient une collection de vos objets connectés (i .e. vos capteurs et vos actionneurs), dont la classe est ‘Things’ (i.e. le nom de votre collection).

**Opérateurs de requête/’query’ MongoDB**

On suppose que notre fonction ‘find’ est la suivante : *db.Things.find(query)       **-** l'attribut projection est ici omis car on veut que tous les champs soient retournés **-***

*Opérateurs de comparaison :*

$eq         valeur d’attribut(s) égale à une valeur donnée

$gt          valeur d’attribut(s) supérieure à une valeur donnée

$gte        valeur d’attribut(s) supérieure ou égale à une valeur donnée

$in          valeur d’attribut(s) égale à n’importe laquelle des valeurs contenues dans un tableau de valeurs donné

$lt           valeur d’attribut(s) inférieure à une valeur donnée.

$lte         valeur d’attribut(s) inférieure ou égale à une valeur donnée

$ne         valeur d’attribut(s) différente d’une valeur donnée

$nin        valeur d’attribut(s) différente de n’importe laquelle des valeurs contenues dans un tableau de valeurs donné

Exemples 

·         Quels sont tous les objets connectés de type capteur   **=>**   *db.Things.find( { thingType: "Sensor" } )* ****-** Ici, on utilise l'expression simplifiée de l'opérateur '$eq' car, en MongoDB, l'expression  { thingType: { $eq: "Sensor" } } est équivalente à l'expression** ***{ thingType: "Sensor" } **-*****

·         Quels sont tous les objets connectés dont la valeur (‘value’) de mesure est supérieure  à 25   **=>**   *db.Things.find( { "usedFor.hasData.hasMeasurement.value": { $gt: 25 } } )*

·         Quels sont tous les objets connectés dont la valeur (‘value’) de mesure est comprise entre 12 et 16   **=>**   *db.Things.find( { "usedFor.hasData.hasMeasurement.value": { $gte: 12, $lte: 16 } } )*

·         Quels sont tous les objets connectés dont la valeur de mesure (‘value’) vaut 39 ou 40   **=>**   *db.Things.find( { "usedFor.hasData.hasMeasurement.value": { $in: [ 39,  40 ] } } )*

*Opérateur d’existence :*

$exists: attribut existe (‘true’) ou non (‘false’) -> revoie tous les documents JSON dont l’attribut existe ou non

 Exemple : quels sont tous les objets connectés de type capteur ayant une position géographique de renseignée   **=>**   *db.Things.find( {  location: { $exists: true } } )*

*Opérateurs logiques :*

$and       ‘et logique’ entre deux clauses de recherche   **=>**   retourne tous les documents JSON qui matchent ces deux clauses en même temps

$or         ‘ou logique’ entre deux clauses de recherche   **=>**   retourne tous les documents JSON qui matchent une de ces deux clauses

$not       inverse d’une clause de recherche   **=>**   retourne tous les documents JSON qui ne matchent pas cette clause

$nor       ‘non logique’ entre deux clauses de recherche   **=>**   retourne tous les documents JSON qui ne matchent pas ces deux clauses en même temps

Exemple : quels sont les capteurs de température dont la valeur est inférieure ou égale à 19°C   **=>**   *db.Things.find( { $and: [ { "usedFor.processID": "* Temperature Measurement*"* *}, { "usedFor.hasData.hasMeasurement.value": { $lte: 19 } } ] } )*

*Opérateurs de tableaux :*

$all                         valeur d’attribut(s) de type tableau qui contient tous les éléments spécifiés dans la clause de recherche

$elemMatch         sélectionne les documents JSON dont la valeur d’attribut(s) de type tableau qui contient un élément matchant toutes les conditions ‘$elemMatch’

$size                       sélectionne les documents JSON dont la valeur d’attribut(s) de type tableau a une taille donnée

Exemple : quels sont les objets connectés fournissant une mesure de température non vide   **=>**   *db.Things.find( { "usedFor.hasData": {  $elemMatch: { name: "temperature", hasMeasurement: { $exists: true } } } } )*

**Opérateurs de ’projection’ MongoDB**

On suppose que notre fonction ‘find’ est la suivante : *db.Things.find (query, projection)*

Exemples :

·         quels sont les capteurs de température dont la valeur est inférieure à 19°C, ne retourner que les attributs Name et Value   **=>**   *db.Things.find( { $and: [ { "usedFor.processID": "* Temperature Measurement*"* *}, { "usedFor.hasData.hasMeasurement.value": { $lte: 19 } } ] }* *, { "usedFor.hasData.name": 1, "usedFor.hasData.hasMeasurement.value": 1 } )*

·         quels sont tous les objets connectés de type capteur, ne pas retourner l’attribut *asData*    **=>**   *db.Things.find( { thingType: "Sensor" }, { "usedFor.hasData": 0 } )*
